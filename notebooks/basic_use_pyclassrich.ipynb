```python
%%time

annotated = classify(params['nap_taskid'])
annotated.head()
```

```python
annotated.shape
```

```python
gnps = Proteosafe(params['gnps_taskid'], 'FBMN')
gnps.get_gnps()

feat = gnps.feat
meta = gnps.meta

meta.head()
```

```python
from pyclassrich.stats import *
```

```python
tabgnps = ndict.tabgnps

tabgnps.head()
```

```python
df = univariate(annotated, tabgnps, 
                params, feat, meta)

df.head()
```

```python
df.columns
```

```python
chemrich = df.copy().loc[~df['cluster.index'].isnull(),
                         ['Identifier', 'class_name', 'InChI', 
                          'pval', 'fchange']]

chemrich.columns = ['Compound_Name', 'Class', 'InChI', 'pvalue', 'foldchange']

clusterdf = class_enrichment(chemrich, cfield='Class', nfield='Compound_Name')
clusterdf.head()
```

```python
pr_lip_loc = df['class_name']=='Prenol lipids'

pr_lip_loc.sum()
```

```python
sum(df.loc[pr_lip_loc, 'pval'] < 0.05)
```

```python
clusterdf2 = clusterdf.sort_values('pvalues').head(10)

impact_plot(clusterdf2, chemrich, cfield='Class', nfield='Compound_Name',
            scaling_factor=20)
```

```python
get_ont_graph(df, params, cfield='Class', nfield='compound names', 
              method='kstest', outdir='.', scaling_factor=10)
```
